```python
# Cell 1: imports, loading and column selection
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/home/abdoulaye/Documents/Orange Digital center/Machine Learning Zoomcamp 2026/car_fuel_efficiency_2026.csv")

# Keep only the columns required by the homework
columns = ['engine_displacement', 'horsepower', 'vehicle_weight',
           'model_year', 'fuel_efficiency_mpg']
df = df[columns]

df.head()
```

```python
# Cell 5: shuffle and split into train/val/test (60%/20%/20%)
def split_data(df, seed):
    n = len(df)
    n_val = int(n * 0.2)
    n_test = int(n * 0.2)
    n_train = n - n_val - n_test

    idx = np.arange(n)
    np.random.seed(seed)
    np.random.shuffle(idx)
    df_shuffled = df.iloc[idx]

    df_train = df_shuffled.iloc[:n_train].copy()
    df_val = df_shuffled.iloc[n_train:n_train + n_val].copy()
    df_test = df_shuffled.iloc[n_train + n_val:].copy()
    return df_train, df_val, df_test

df_train, df_val, df_test = split_data(df, seed=42)
y_train = df_train['fuel_efficiency_mpg'].values
y_val = df_val['fuel_efficiency_mpg'].values

print(len(df_train), len(df_val), len(df_test))
```

```python
# Cell 6: linear regression (r=0 means no regularization) and RMSE
features = ['engine_displacement', 'horsepower', 'vehicle_weight', 'model_year']

def prepare_X(df, fill_value):
    return df[features].fillna(fill_value).values

def train_linear_regression(X, y, r=0.0):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX = XTX + r * np.eye(XTX.shape[0])
    XTX_inv = np.linalg.inv(XTX)

    w_full = XTX_inv.dot(X.T).dot(y)
    return w_full[0], w_full[1:]

def rmse(y, y_pred):
    return np.sqrt(((y - y_pred) ** 2).mean())
```

```python
# Cell 7: compare filling NAs with 0 vs the mean (mean computed on train only)
mean_hp = df_train['horsepower'].mean()

for name, fill in [('With 0', 0), ('With mean', mean_hp)]:
    X_train = prepare_X(df_train, fill)
    w0, w = train_linear_regression(X_train, y_train)

    X_val = prepare_X(df_val, fill)
    y_pred = w0 + X_val.dot(w)

    print(name, "-> RMSE:", round(rmse(y_val, y_pred), 2))
```

```python
# Cell 8: try different regularization values, NAs filled with 0
X_train = prepare_X(df_train, 0)
X_val = prepare_X(df_val, 0)

for r in [0, 0.01, 0.1, 1, 5, 10, 100]:
    w0, w = train_linear_regression(X_train, y_train, r=r)
    y_pred = w0 + X_val.dot(w)
    print(r, "-> RMSE:", round(rmse(y_val, y_pred), 2))
```

```python
# Cell 9: RMSE on validation for seeds 0 to 9, NAs filled with 0, no regularization
scores = []

for seed in range(10):
    df_train_s, df_val_s, df_test_s = split_data(df, seed=seed)

    X_train_s = prepare_X(df_train_s, 0)
    y_train_s = df_train_s['fuel_efficiency_mpg'].values
    w0, w = train_linear_regression(X_train_s, y_train_s)

    X_val_s = prepare_X(df_val_s, 0)
    y_val_s = df_val_s['fuel_efficiency_mpg'].values
    y_pred = w0 + X_val_s.dot(w)

    scores.append(rmse(y_val_s, y_pred))

print("Scores:", [round(s, 3) for s in scores])
print("Std:", round(np.std(scores), 3))
```

```python
# Cell 10: seed 9, combine train + val, fill NAs with 0, train with r=0.001, evaluate on test
df_train_9, df_val_9, df_test_9 = split_data(df, seed=9)

# Combine train and validation sets
df_full_train = pd.concat([df_train_9, df_val_9])

X_full_train = prepare_X(df_full_train, 0)
y_full_train = df_full_train['fuel_efficiency_mpg'].values
w0, w = train_linear_regression(X_full_train, y_full_train, r=0.001)

X_test = prepare_X(df_test_9, 0)
y_test = df_test_9['fuel_efficiency_mpg'].values
y_pred = w0 + X_test.dot(w)

print("Test RMSE:", rmse(y_test, y_pred))
```
