## Cell 6 — Question 6: Median horsepower before/after fill

```python
# Calculate the median horsepower before filling missing values
median_before = df['horsepower'].median()
print("Median before:", median_before)

# Find the most frequent horsepower value (mode)
mode_hp = df['horsepower'].mode()[0]
print("Mode:", mode_hp)

# Fill missing horsepower values with the mode
df['horsepower'] = df['horsepower'].fillna(mode_hp)

# Calculate the median horsepower after filling missing values
median_after = df['horsepower'].median()
print("Median after:", median_after)

# Compare the two median values
if median_after > median_before:
    print("Yes, it increased")
elif median_after < median_before:
    print("Yes, it decreased")
else:
    print("No")
```

```python
# Q7 — Sum of weights

# Select cars from Asia
cars_from_asia = df[df['origin'] == 'Asia']

# Keep only vehicle_weight and model_year
cars_from_asia = cars_from_asia[['vehicle_weight', 'model_year']]

# Select first 7 rows
cars_from_asia_first_seven = cars_from_asia.iloc[:7, :]

# Convert to NumPy array X
X = cars_from_asia_first_seven.to_numpy()

# Compute XTX
XT = X.T
XTX = np.dot(XT, X)

#  Invert XTX
XTX_inv = np.linalg.inv(XTX)

#  Create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# Compute weights
w = np.dot(np.dot(XTX_inv, XT), y)

#Sum of weights
print("Weights:", w)
print("Sum:", np.sum(w))
```
